attaching labels to schemes

```python
import pandas as pd

dataset = {
    "scheme1_bow_model": "data/scheme1/scheme1_bow.csv",
    "scheme2_bow_model": "data/scheme2/scheme2_bow.csv",
    "scheme3_bow_model": "data/scheme3/scheme3_bow.csv",
    "scheme1_glove_model": "data/scheme1/scheme1_glove.csv",
    "scheme2_glove_model": "data/scheme2/scheme2_glove.csv",
    "scheme3_glove_model": "data/scheme3/scheme3_glove.csv",
}

def attach_labels_to_datasets(dataset_dict, labels_series, label_col="majority_vote_final"):
    print("Attaching labels to datasets...")

    labels = labels_series.values

    for name, path in dataset_dict.items():
        print(f"\nProcessing {name} -> {path}")

        df = pd.read_csv(path)

        if len(df) != len(labels):
            raise ValueError(
                f"Row mismatch in {name}: dataset={len(df)} vs labels={len(labels)}"
            )

        df[label_col] = labels

        out_path = path.replace(".csv", "_labeled.csv")
        df.to_csv(out_path, index=False)

        print(f"Saved: {out_path}")
if __name__ == "__main__":

    labels = imp_df_extracted["majority_vote_final"]

    attach_labels_to_datasets(dataset, labels)
```

```python
import pandas as pd
import nltk
from nltk.corpus import sentiwordnet as swn
from nltk.corpus import wordnet as wn
from nltk.tokenize import word_tokenize
import os

nltk.download('punkt')
nltk.download('sentiwordnet')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')


class BingLiuClassifier:
    def __init__(self, pos_file, neg_file):
        self.pos_words = self._load_lexicon(pos_file)
        self.neg_words = self._load_lexicon(neg_file)
        self.negation_words = {
            "not", "no", "never", "n't", "hardly",
            "barely", "neither", "nor", "none"
        }

    def _load_lexicon(self, file_path):
        words = set()
        with open(file_path, 'r', encoding='iso-8859-1') as f:
            for line in f:
                if not line.startswith(';') and line.strip():
                    words.add(line.strip().lower())
        return words

    def analyze(self, text):
        if pd.isna(text):
            return "Neutral"

        tokens = word_tokenize(str(text).lower())
        score = 0

        for i, word in enumerate(tokens):
            sentiment = 0

            if word in self.pos_words:
                sentiment = 1
            elif word in self.neg_words:
                sentiment = -1

            if i > 0 and tokens[i - 1] in self.negation_words:
                sentiment *= -1

            score += sentiment

        return "Positive" if score > 0 else "Negative" if score < 0 else "Neutral"

class SentiWordNetClassifier:
    def get_wordnet_pos(self, tag):
        if tag.startswith('J'):
            return wn.ADJ
        elif tag.startswith('V'):
            return wn.VERB
        elif tag.startswith('N'):
            return wn.NOUN
        elif tag.startswith('R'):
            return wn.ADV
        return None

    def analyze(self, text):
        if pd.isna(text):
            return "Neutral"

        tokens = word_tokenize(str(text))
        tagged = nltk.pos_tag(tokens)

        score = 0

        for word, tag in tagged:
            wn_tag = self.get_wordnet_pos(tag)
            if wn_tag:
                synsets = list(swn.senti_synsets(word, wn_tag))
                if synsets:
                    s = synsets[0]
                    score += (s.pos_score() - s.neg_score())

        if score > 0.01:
            return "Positive"
        elif score < -0.01:
            return "Negative"
        return "Neutral"

def run_all_schemes():

    datasets = {
        "scheme1": "data/scheme1/cleaned_comments1.csv",
        "scheme2": "data/scheme2/cleaned_comments2.csv",
        "scheme3": "data/scheme3/cleaned_comments3.csv",
    }

    positive_lex = "positive-words.txt"
    negative_lex = "negative-words.txt"

    bing = BingLiuClassifier(positive_lex, negative_lex)
    swn_model = SentiWordNetClassifier()

    for scheme, input_file in datasets.items():
        print(f"\nProcessing {scheme}...")

        if not os.path.exists(input_file):
            print(f"Missing file: {input_file}")
            continue

        df = pd.read_csv(input_file)

        if "clean_comment" not in df.columns:
            print(f"Skipping {scheme}: no clean_comment column")
            continue

        df["clean_comment"] = df["clean_comment"].fillna("").astype(str)

        print("Running Bing Liu...")
        df["bing_liu_label"] = df["clean_comment"].apply(bing.analyze)

        print("Running SentiWordNet...")
        df["sentiwordnet_label"] = df["clean_comment"].apply(swn_model.analyze)

        output_file = f"data/{scheme}/{scheme}_lexical_labeled.csv"
        df.to_csv(output_file, index=False)

        print(f"Saved → {output_file}")

if __name__ == "__main__":
    run_all_schemes()
```

```python
from sklearn.metrics import confusion_matrix, f1_score, accuracy_score

datasets = [
    "data/scheme1/scheme1_lexical_labeled.csv",
    "data/scheme2/scheme2_lexical_labeled.csv",
    "data/scheme3/scheme3_lexical_labeled.csv"
]

results = []

for path in datasets:

    df = pd.read_csv(path)

    scheme_name = path.split("/")[-1].replace("_lexical_labeled.csv", "")

    y_true = df["majority_vote_final"].astype(str).str.lower().str.strip()
    y_true = y_true.replace({"negtaive": "negative"})
    y_pred_blu = df["bing_liu_label"].astype(str).str.lower().str.strip()
    y_pred_senti = df["sentiwordnet_label"].astype(str).str.lower().str.strip()

    print(f"\n{scheme_name}")
    print("TRUE LABELS:", y_true.unique())
    print("PRED LABELS blu:", y_pred_blu.unique())
    print("PRED LABELS senti:", y_pred_senti.unique())

    f1_blu = f1_score(y_true, y_pred_blu, average="weighted")
    acc_blu = accuracy_score(y_true, y_pred_blu)
    cm_blu = confusion_matrix(y_true, y_pred_blu)

    results.append({
        "scheme": scheme_name,
        "model": "BingLiu",
        "f1": f1_blu,
        "accuracy": acc_blu,
        "confusion_matrix": cm_blu
    })

    f1_senti = f1_score(y_true, y_pred_senti, average="weighted")
    acc_senti = accuracy_score(y_true, y_pred_senti)
    cm_senti = confusion_matrix(y_true, y_pred_senti)

    results.append({
        "scheme": scheme_name,
        "model": "SentiWordNet",
        "f1": f1_senti,
        "accuracy": acc_senti,
        "confusion_matrix": cm_senti
    })

output_path = "results/lexical_models_report.txt"

with open(output_path, "w", encoding="utf-8") as f:
    f.write("LEXICAL MODELS BENCHMARK REPORT\n")
    f.write("=" * 60 + "\n\n")

    for r in results:
        f.write(f"Scheme: {r['scheme']}\n")
        f.write(f"Model: {r['model']}\n")
        f.write(f"F1 Score (weighted): {r['f1']:.4f}\n")
        f.write(f"Accuracy: {r['accuracy']:.4f}\n")
        f.write("Confusion Matrix:\n")
        f.write(str(r["confusion_matrix"]))
        f.write("\n" + "-" * 50 + "\n\n")

print(f"Saved report to: {output_path}")
```
